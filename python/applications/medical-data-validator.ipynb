**In this project, we will write a program that checks whether the health data stored in a python dictionary is in the desired format.**

```python
# First of all, we need to import some modules that we will use in our code.
import re
```

```python
# We need some data to test our code. We will store dictionaries within a list.

medical_records = [
    {
        'patient_id': 'P1001',
        'age': 34,
        'gender': 'Female',
        'diagnosis': 'Hypertension',
        'medications': ['Lisinopril'],
        'last_visit_id': 'V2301',
    },
    {
        'patient_id': 'p1002',
        'age': 47,
        'gender': 'male',
        'diagnosis': 'Type 2 Diabetes',
        'medications': ['Metformin', 'Insulin'],
        'last_visit_id': 'v2302',
    },
    {
        'patient_id': 'P1003',
        'age': 29,
        'gender': 'female',
        'diagnosis': 'Asthma',
        'medications': ['Albuterol'],
        'last_visit_id': 'v2303',
    },
    {
        'patient_id': 'p1004',
        'age': 56,
        'gender': 'Male',
        'diagnosis': 'Chronic Back Pain',
        'medications': ['Ibuprofen', 'Physical Therapy'],
        'last_visit_id': 'V2304',
    }
]
```

```python
# Our data is ready now, we can start validating now. First of all, let's check whether the value entered for each key is in the correct format.

def find_invalid_records(patient_id, age, gender, diagnosis, medications, last_visit_id):

  constraints = {
      'patient_id': (isinstance(patient_id, str) and re.fullmatch(r'p\d+', patient_id, re.IGNORECASE)), # We used a regex to check string format.
      'age': (isinstance(age, int) and age >= 18),
      'gender': (isinstance(gender, str) and gender.lower() in ('male','female')),
      'medications': (isinstance(medications, list) and all([isinstance(i, str) for i in medications])),
      'last_visit_id': (isinstance(last_visit_id, str) and re.fullmatch(r'v\d+', last_visit_id, re.IGNORECASE))
  }

  return [key for key, value in constraints.items() if not value] # We return the keys of values that do not satisfy the constraints we want.
```

```python
# Now let's perform general checks using our constraints.

def validate(data):

  is_sequence = isinstance(data, (list, tuple))

  if not is_sequence:
    print('Invalid format: expected a list or tuple.')
    return False

  is_invalid = False
  key_set = set(['patient_id', 'age', 'gender', 'medications', 'last_visit_id'])

  for index, dictionary in enumerate(data):

    # We checked whether the values in the list were dictionaries or not.
    if not isinstance(dictionary, dict):
      print(f'Invalid format: expected a dictionary at position {index}.')
      is_invalid = True
      continue # If our data has a data type other than a dictionary, we used 'continue' to prevent the code from continuing and causing an error.

    # We checked whether all the keys in our dictionary were the keys we wanted.
    if set(dictionary.keys()) != key_set:
      print(f'Invalid format: {dictionary} at position {index} has missing and/or invalid keys.')
      is_invalid = True
      continue # ** operator matches each key in the dictionary with the parameter of the same name of the function and transfers the value of this key
               # to the corresponding parameter. Therefore, if one of the keys comes with the wrong name, we will put continue so that the code
               # continues and does not give an error.

    invalid_records = find_invalid_records(**dictionary)

    for invalid_key in invalid_records:
      print(f"Unexpected format '{invalid_key}: {dictionary[invalid_key]}' at position {index}.")
      is_invalid = True

  if is_invalid:
    return False

  print('Valid format.')
  return True
```
