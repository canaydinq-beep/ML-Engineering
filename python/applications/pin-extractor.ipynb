**Problem:** The nth digit of the pin is hidden as the length of the nth word in the nth line.

**Purpose:** Finding the hidden pins within the poem.

**Edge Cases:** The line of the poem could be shorter than expected.

```python

def pin_extractor(poems):
  pins = []

  for poem in poems:
    pin = ''
    lines = poem.split('\n')

    for line_index, line in enumerate(lines, 1):
      words = line.split(' ')

      if len(words) >= line_index:
        pin += str(len(words[line_index]))
      else:
        pin += 0

    pins.append(pin)

  return pins
```

**Sample Poems for Use**
