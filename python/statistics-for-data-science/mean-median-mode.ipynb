# **Calculation of Mean, Median, and Mode in Python**

**What is the mean and where is it used?**

It is the arithmetic central value obtained by summing all values in a dataset and dividing by the numbers of observations.


*   To summarize the overall trend in data that is symmetric and normally distributed,

*   During the steps for filling in numerical missing data (mean imputation),

*   It is used in standardization (Z-score scaling) processes and regression metrics such as MSE.

**What is the median and where is it used?**

It is the value in the exact middle of a data set sorted from smallest to largest; if the number of data points is even, it is the arithmetic mean of the two middle values.


*   To determine the non-outlying center value in data sets containing outliers and skewed distributions (e.g., income levels or house prices),

*   Resistant to extreme values in missing data filling operations,

*   It is used in box-plot analysis and robust scaling methods.

**What is the mode and where is it used?**

It is the value that occurs most frequently in a dataset.


*   To determine the central tendency of non-numerical categorical data,

*   In the steps of filling in missing data in categorical columns with the most common class,

*   Used in classification problems to select class with most votes (e.g., voting in random forest or k-NN algorithms) and in segmentation analyses.
























```python
import numpy as np
import pandas as pd
import statistics as st
```

**Sample Dataset**

```python
sample_data = [742, 742, 89, 415, 932, 18, 604, 251, 877, 33, 512, 164, 789, 401, 67, 955, 328, 120, 843, 619, 45, 731, 290, 564, 982, 15, 408, 861]
```

**Mean - Median - Mode Calculations with NumPy**

```python
# Mean Calculation
np.mean(sample_data)

# Median Calculation
np.median(sample_data)

# Mode Calculation
values, frequencies = np.unique(sample_data, return_counts = True)
max_frequency = np.argmax(frequencies)
_mode = values[max_frequency]
```

```python
# First, we need to convert this data, which is in list form, into a Pandas DataFrame.
sample_data_df = pd.DataFrame({'values':sample_data})
```

```python
# Mean Calculation
sample_data_df['values'].mean()

# Median Calculation
sample_data_df['values'].median()

# Mode Calculation
sample_data_df['values'].mode()
```

```python
# Mean Calculation
st.mean(sample_data)

# Median Calculation
st.median(sample_data)

# Mode Calculation
st.mode(sample_data)
```

**Mean - Median - Mode Calculation with Python Functions** **(args)**

```python
# When a defining a function in Python, the *args parameter takes the function arguments wrapped in a tuple.
# In other words, all the values you want to pass into function are written into a tuple, and the function executes the values within the tuple.

# func1(1,2,3,4) -> (1,2,3,4) -> numbers inside a tuple

# func1((1,2,3,4)) -> ((1,2,3,4)) -> tuple inside a tuple

# func1([1,2,3,4]) -> ([1,2,3,4]) -> list inside a tuple
```

```python
# Mean Calculation
def mean_calc(*args):

  if len(args) == 1 and isinstance(args[0], (list, tuple)):
    args = args[0]

  if len(args) == 0:
    return None

    return sum(args) / len(args)
```

```python
# Median Calculation
def median_calc(*args):

  if len(args) == 1 and isinstance(args[0], (list, tuple)):
    args = args[0]

  if len(args) == 0:
    return None

  sorted_data = sorted(args)
  mid = len(sorted_data) // 2

  if len(sorted_data) % 2 != 0:
    return sorted_data[mid]

  else:
    return (sorted_data[mid] + sorted_data[mid - 1]) / 2
```

```python
# Mode Calculation

def mode_calc(*args):

  if len(args) == 1 and isinstance(args[0], (list, tuple)):
    args = args[0]

  if len(args) == 0:
    return None

  frequencies = {}

  for arg in args:
    if arg in frequencies:
      frequencies[arg] += 1
    else:
      frequencies[arg] = 1

  val = [num for num, frequency in frequencies.items() if frequency == max(frequencies.values())]
  frequency = [frequency for num, frequency in frequencies.items() if frequency == max(frequencies.values())]

  return val, frequency
```
